#Import Libraries

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

#Load the Iris dataset from seaborn

```python
iris = sns.load_dataset('iris')
```

```python
print("Dataset Shape:", iris.shape)
```

```python
print("\nStatistical Description:")
print(iris.describe())
```

#Scatter Plot: Sepal Length vs Petal Length by Species

```python
plt.figure(figsize=(8, 6))  # Create a new figure with a width of 8 inches and height of 6 inches for better readability and control over plot size.

sns.scatterplot(
    data=iris,                  # Use the 'iris' DataFrame as the data source for the plot.
    x='sepal_length',           # Set the x-axis to represent the 'sepal_length' column.
    y='petal_length',           # Set the y-axis to represent the 'petal_length' column.
    hue='species',              # Differentiate the points based on the 'species' column using color.
    palette='Set2'              # Use the 'Set2' color palette for a clean, visually distinct appearance.
)

plt.title("Sepal Length vs Petal Length")  # Set the title of the plot to describe what is being visualized.

plt.grid(True)  # Enable grid lines in the background to improve readability of data points.

plt.show()  # Render and display the final plot.
```

# Histograms for each feature

```python
iris.hist(
    edgecolor='black',     # Adds a black border around each bar in the histogram for better visual separation.
    linewidth=1.2,         # Sets the thickness of the edge lines (bar borders) to 1.2 points for enhanced clarity.
    figsize=(12, 8)        # Specifies the overall figure size as 12 inches wide and 8 inches tall for a spacious layout.
)
plt.suptitle("Feature Distributions", fontsize=16)
# Adds a main title above all subplots in the figure to describe what the collection of histograms represents.
# 'fontsize=16' increases the font size to make the title more prominent.
plt.tight_layout()
# Adjusts subplot spacing automatically so that axis labels and titles do not overlap — especially important with multiple plots.
plt.show()
# Renders and displays the entire figure with all histograms.
```

# Box Plots to identify outliers

```python
plt.figure(figsize=(12, 6))
# Creates a new figure that is 12 inches wide and 6 inches tall.
# This defines the overall canvas size where all the box plots will be drawn.
for idx, column in enumerate(iris.columns[:-1]):
    # Iterates over the first four columns (numerical features) in the dataset.
    # 'enumerate()' provides both the index (idx) and the column name.
    # 'iris.columns[:-1]' excludes the last column ('species'), which is categorical.
    plt.subplot(1, 4, idx + 1)
    # Creates a 1-row, 4-column subplot layout, and selects the (idx + 1)-th subplot.
    # This ensures each numerical feature gets its own plot placed side-by-side.
    sns.boxplot(y=iris[column], color='skyblue')
    # Draws a vertical box plot for the given numerical column.
    # Box plots show median, quartiles, and potential outliers.
    # 'skyblue' sets the box color to a pleasant blue for visual clarity.
    plt.title(column)
    # Sets the title of the current subplot to the name of the column (feature being plotted).
plt.tight_layout()
# Automatically adjusts spacing between subplots so titles, axes, and labels don’t overlap.
plt.show()
# Renders and displays the entire row of box plots on screen.
```
